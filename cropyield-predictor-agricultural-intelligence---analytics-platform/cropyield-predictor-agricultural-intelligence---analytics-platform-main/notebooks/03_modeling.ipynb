# CropYield Predictor - Model Training and Evaluation

This notebook documents the verified model comparison, tuning evidence and serialized pipeline inference artifact included in the submission.

```
import json
import joblib
import pandas as pd
comparison = pd.DataFrame(json.load(open('../models/model_comparison.json')))
metrics = json.load(open('../models/evaluation_metrics.json'))
comparison
```

```
pipeline = joblib.load('../models/crop_yield_model.joblib')
sample = {'Area':'India','Item':'Rice, paddy','Year':2008,'average_rain_fall_mm_per_year':1200.0,'pesticides_tonnes':40000.0,'avg_temp':26.5}
prediction = float(pipeline.predict(sample)[0])
print('Final model:', metrics['final_model'])
print('Held-out R2:', metrics['test_r2'])
print('Held-out MAE:', metrics['test_mae'])
print('Held-out RMSE:', metrics['test_rmse'])
print('Sample prediction (hg/ha):', round(prediction,2))
```
